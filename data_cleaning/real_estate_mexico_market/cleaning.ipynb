```python
#cleaning the dataset
df1 = (
    df
    .dropna()  # drops all rows with missing values in at least one of the columns
    .drop_duplicates() # drops all duplicates
    .assign(
        price_usd = lambda x:
          x["price_usd"]
          .str.replace("$", "", regex=False) #without regex=False python treats it as a regular expression
          .str.replace(",", "", regex=False) # we use regex=False to speficy that $ is not a regualar expression because in regex $ is end of string
          .astype(float)
    )
  )
df1.isnull().sum()
```

```python
#checking if the first dataset is clean
df1["price_usd"].dtype #checks for the datatype of price_usd column
```

## Cleaning the rest of the Dataset and combining them

```python
#Define the path of the dataset2
path_2 = "./data/mexico-real-estate-2.csv"
df2 = pd.read_csv(path_2)
df2
```

```python
#cleaning df2
df2 = (
    pd.read_csv(path_2)
    .assign(
        price_usd = lambda x: x["price_mxn"].div(19)
    )
    .drop(columns=["price_mxn"])
    .dropna()
)
print(f"df2 type: {type(df2)}")
print(f"df2 type: {df2.shape}")
```

```python
#verifiy df2 and comparing with df1
df2.columns.tolist()
df1.columns.tolist()
df2.isnull().sum()      #check to ensure that df2 has no null or missing vallue
df2["price_usd"].dtype  # check to ensure the datatype is float
```

```python
#cleaning df3 the third dataset
path_3 = "./data/mexico-real-estate-3.csv"
df3 =(
    pd.read_csv(path_3)
    .dropna()
    .assign(
        lat= lambda x: x["lat-lon"]
            .str.split(",", expand=True)[0]
            .astype(float),

        lon= lambda x: x["lat-lon"]
                    .str.split(",", expand=True)[1]
                    .astype(float),
        state= lambda x: x["place_with_parent_names"]
        .str.split("|", expand=True)[2]
    )
    .drop(columns=["place_with_parent_names", "lat-lon"])
)
# Print object type, shape, and head
print("df3 type:", type(df3))
print("df3 shape:", df3.shape)
df3.head()
```

```python
# Check if all column sets are identical
columns_match = set(df1.columns) == set(df2.columns) == set(df3.columns)

if columns_match:
    print("✅ All DataFrames have the same columns.")
else:
    print("❌ Column mismatch detected!")
    # Optional: See the difference
    print("df1 vs df2 diff:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3 diff:", set(df2.columns) ^ set(df3.columns))
```
